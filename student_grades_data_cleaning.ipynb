```python
df1 = pd.read_csv("grades.csv")
```

```python
duplicates = df1[df1.duplicated()]
duplicates
```

```python
# 8 dropping duplicate rows
```

```python
df1 = df1.drop_duplicates()

# Verifying duplicates are removed
df1[df1.duplicated()]
```

```python
# 9 Finding missing values
```

```python
missing = df1.isnull().any(axis=1)
df1[missing]
```

```python
# 10 dropping rows with missing values
```

```python
df1 = df1.dropna()

# Verifying that no missing values remain
df1.isnull().any(axis=1)
```
